Merging all the tables

```python
data=sales.merge(products,on='product_id')
data=data.merge(stores,on='store_id')
data=data.merge(calendar,left_on='order_date',right_on='date')
data=data.merge(customers,on='customer_id')
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
year_data=data.groupby('year').agg({'order_id':'nunique','quantity':sum,'revenue':sum,'profit':sum}).reset_index()
year_data.columns=['Year','Number_of_orders','Total_Quantity','Total_Revenue','Total_Profit']
year_data.head()
```

```python
month_data=data.groupby('month')
month_data=month_data.agg({'order_id':'nunique','quantity':sum,'revenue':sum,'profit':sum}).reset_index()
month_data.columns=['Month','Number_of_orders','Total_Quantity','Total_Revenue','Total_Profit']
month_data.head(12)
```

```python
brand_data=data.groupby('brand').agg({'order_id':'nunique','quantity':sum,'revenue':sum,'profit':sum}).reset_index()
brand_data.columns=['Brand','Number_of_orders','Total_Quantity','Total_Revenue','Total_Profit']
```

```python
plt.figure(figsize=(6,5))
sns.barplot(y=brand_data['Number_of_orders'],x=brand_data['Brand'])
plt.title('Total orders per Brand')
plt.show()
plt.figure(figsize=(6,5))
sns.barplot(y=brand_data['Total_Quantity'],x=brand_data['Brand'])
plt.title('Total Quantity per Brand')
plt.show()
plt.figure(figsize=(6,5))
sns.barplot(y=brand_data['Total_Revenue'],x=brand_data['Brand'])
plt.title('Total Revenue per Brand')
plt.show()
plt.figure(figsize=(6,5))
sns.barplot(y=brand_data['Total_Profit'],x=brand_data['Brand'])
plt.title('Total Profit per Brand')
plt.show()
```

```python
store_type_data=data.groupby('store_type').agg({'order_id':'nunique','quantity':sum,'revenue':sum,'profit':sum}).reset_index()
store_type_data.columns=['Store_type','Number_of_orders','Total_Quantity','Total_Revenue','Total_Profit']
store_type_data.head()
```

```python
plt.figure(figsize=(6,5))
sns.barplot(y=store_type_data['Number_of_orders'],x=store_type_data['Store_type'])
plt.title('Total orders per Store type')
plt.show()
plt.figure(figsize=(6,5))
sns.barplot(y=store_type_data['Total_Quantity'],x=store_type_data['Store_type'])
plt.title('Total Quantity per Store type')
plt.show()
plt.figure(figsize=(6,5))
sns.barplot(y=store_type_data['Total_Revenue'],x=store_type_data['Store_type'])
plt.title('Total Revenue per Store type')
plt.show()
plt.figure(figsize=(6,5))
sns.barplot(y=store_type_data['Total_Profit'],x=store_type_data['Store_type'])
plt.title('Total Profit per Store type')
plt.show()
```

```python
city_data=data.groupby('city').agg({'order_id':'nunique','quantity':sum,'revenue':sum,'profit':sum}).reset_index()
city_data.columns=['City','Number_of_orders','Total_Quantity','Total_Revenue','Total_Profit']
```
